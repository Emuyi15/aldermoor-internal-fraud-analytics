```python
peer_counts = [28, 35, 31, 147, 29, 33, 40, 26, 32,]
#calculate_median(peer_counts)
# assert peer_counts == [28, 35, 31, 147, 29, 33, 40, 26, 32]

def calculate_median(values: list) -> float:

    '''
        calculate the median of any given list of numbers
        returns a float

        '''
    if not values:
        raise ValueError(f"Cannot calculate the median of an empty list")

    sorted_values = sorted(values)
    n_values = len(sorted_values)
    middle = n_values // 2
    
    if n_values % 2 == 1:
        return sorted_values[middle]

    return (sorted_values[middle - 1] + sorted_values[middle]) / 2

assert calculate_median([2, 5, 3]) == 3.0
assert calculate_median([2, 5, 3, 10]) == 4.0

peer_median = calculate_median(peer_counts)
print(peer_median)

# try the empty list valueerror catch

try:
    test_peer_counts = []
    peer_median = calculate_median(test_peer_counts)
except ValueError as e:
    print(f"Raised as expected: {e}")

else:
    raise AssertionError("calculate_median([]) should have raised ValueError")
```

```python
# Part 3: parse_count(raw_value: str) -> int
def parse_count(raw_value: str) -> int:

    """
    standardize input and change str to int
    """        
    parse_value = (
        raw_value.strip()
    )
    parsed_value = int(parse_value)

    if parsed_value < 0:
        raise ValueError(f"The number is a negative {parsed_value!r}")

    return parsed_value

assert parse_count("42") == 42
assert parse_count(" 38") == 38
assert parse_count("51\n") == 51
assert parse_count("0") == 0
```

```python
for bad_value in ["N/A", "", "-4", "14O"]:
    try:
        parse_count(bad_value)
    except ValueError as e:
        print(f"{bad_value!r} raised")
    else:
        raise AssertionError(f"parse_count({bad_value}) should have raised ValueError")

print("parse count: all tests passed")
```

```python
# Part 4: Quarantine bad rows
raw_daily_counts = [
    ("E00417", "42"),
    ("E00588", " 38"),
    ("E00612", "51\n"),
    ("E00655", "N/A"),
    ("E00701", "-4"),
    ("E00733", "47"),
    ("E00790", ""),
    ("E00812", "55"),
    ("E00845", "14O"),
]

valid_counts = {}
rejected_list = []
for employee_id, count in raw_daily_counts:
    try:
        valid_counts[employee_id] = parse_count(count)
    except ValueError as e:
        rejected_list.append((employee_id, count, str(e)))

print(valid_counts)

for employee_id, count, reason in rejected_list:
    print(f"\n {employee_id}, {count}, {reason}")
```
