```python
raw_role = "  BRANCH   Adviser\n"

raw_role = " ".join(
    raw_role
    .lower()
    .split()
)

print(raw_role)
print(len(raw_role))
```

```python
# clean several values

raw_department = "Fraud  Operations "
raw_system = "PAYMENTS-HUB"
raw_location = " leeds briggate"
raw_action = "View_Account"


def clean_several(text):
    text = "_".join(text.replace("-", " ").lower().split())

    return text

clean_department = clean_several(raw_department)
clean_system = clean_several(raw_system) 
clean_location = clean_several(raw_location)
clean_action = clean_several(raw_action)


print(clean_department)
print(clean_system)
print(clean_location)
print(clean_action)
```
