```python
# Part 2: The hidden input
EXCESSIVE_ACCESS_MULTIPLIER = 3
peer_median = 32

def is_excessive(unique_customers):
    return unique_customers / peer_median > EXCESSIVE_ACCESS_MULTIPLIER

print(is_excessive(110))
peer_median = 40
print(is_excessive(110))

# with peer_median 32 = True
# with peer_median 40 = False

# explanation: the function has a hidden input
# as a result, mutatable global variables would affect the result of 
# the function when they're changed
```

```python
# Part 3: UnboundLocalError
alerts_raised = 0

def raise_alert(employee_id):
    print(f"Alert raised for {employee_id}")
    alerts_raised = alerts_raised + 1

raise_alert("E00417")

# Unbound local error

# this happens because python assumes alert_raised is local to the function
# However, when it tries to read the local before it has a variable, it failed
```

```python
# Part 3: UnboundLocalError
alerts_raised = 0

def raise_alert(employee_id):
    print(f"Alert raised for {employee_id}")
    global alerts_raised

    alerts_raised += 1

raise_alert("E00417")
```
