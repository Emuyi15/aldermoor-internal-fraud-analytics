```python
# Part 1: Unique and repeated access
# Customer IDs E00417 viewed on Tuesday morning, in order:
customers_viewed = ["C10231", "C10877", "C10231", "C22019", "C10231", "C30551", "C10877", "C10231"]

# Calculate and print:
# The total number of views
# The number of unique customers
# The repeat-view ratio (total views ÷ unique customers)
# How many times C10231 was viewed

total_views = len(customers_viewed)
unique_customers_viewed = len(set(customers_viewed))
repeat_view_ratio = total_views / unique_customers_viewed
n_times_a_customer_was_viewed = customers_viewed.count("C10231")

print(f" Total Views: {total_views} ")
print(f" Unique customers viewed: {unique_customers_viewed} ")
print(f" repeat_view_ratio: {repeat_view_ratio} ")
print(f" Number of times viewed: {n_times_a_customer_was_viewed}")

```

```python
# Part 3: Shared customers between two employees
# Customers accessed this week by two Leeds Briggate advisers:

e00417_customers = {"C10231", "C10877", "C22019", "C30551", "C40112"}
e00588_customers = {"C10231", "C22019", "C30551", "C51876", "C60233", "C70045"}

# Calculate and print:
#The customers both accessed, and how many
#The customers either accessed, and how many
#The overlap score: shared count ÷ combined count

#That overlap score is called Jaccard similarity.
#It runs from 0 (nothing in common) to 1 (identical sets).
#We'll use it as a collusion feature later.

# customers both accessed
both_accessed = e00417_customers & e00588_customers
total_both_accessed = len(both_accessed)
print(f" Both of them accessed {both_accessed} which is a total of {total_both_accessed}")

# customers either accessed and how many
customers_either_accessed = e00417_customers | e00588_customers
total_customers_either_accessed = len(customers_either_accessed)
print(f" Either of them accessed {customers_either_accessed} which is a total of {total_customers_either_accessed}")

# The overlap score (jaccard similarity)
jaccard_similarity = total_both_accessed / total_customers_either_accessed
print(f"Jaccard Similarity: {jaccard_similarity:.2f}")
```
