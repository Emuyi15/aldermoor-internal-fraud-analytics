# LIST AND TUPLES

```python
# daily access
daily_access_events = [42, 38, 51, 47, 55, 0, 0]

# total event
total_event = sum(daily_access_events)

#busiest day count
busiest_day_count = max(daily_access_events)

# average per day
average_access_per_day = sum(daily_access_events)/len(daily_access_events)

# number of days with no activity
n_days_with_no_activity = daily_access_events.count(0)

print(total_event)
print(busiest_day_count)
print(f'{average_access_per_day:.2f}')
print(n_days_with_no_activity)
```

```python
# Part 2: Weekdays and weekends
# weekday
weekday = daily_access_events[:5]

# weekend
weekend = daily_access_events[5:]
print(weekday)
print(weekend)
```

```python
# Part 3: A late record and a correction
# update event log with new record
daily_access_events.append(44)

print(daily_access_events)

# correct event log with new findings
daily_access_events[2] = 49

print(daily_access_events)

# We are able to do this because list are mutable while strings are immutable
```

```python
# Part 4: The sorting trap
peer_weekly_customers = [28, 35, 31, 147, 29, 33, 40, 26, 32]

sorted_customers = sorted(peer_weekly_customers)

print(peer_weekly_customers)
print(sorted_customers)

# it returned None because .sort is a method that sorts the original list and should have use sorted()
```

```python
# Part 7: The aliasing trap

original_counts = [42, 38, 51]
working_counts = original_counts
working_counts.append(99)
print(original_counts)

# The output happened so because working counts is just like another sticker stuck on a container.
#Like giving the list another name

# the fix
original_counts.remove(99)
working_counts = original_counts.copy()
working_counts.append(99)
print(f'corrected working count{working_counts}')
print(f'corrected original counts {original_counts}')
```
