```python
# classify_access_period(hour: int, day: str) -> str

# Aldermoor's normal working hours are 08:00 to 18:00: from hour 8, up to but not including hour 18.

WORKING_HOURS_START = 8
WORKING_HOURS_END = 18
WEEKEND_DAYS = {"saturday", "sunday"}

access_events = [
    (9, "Monday"),
    (21, "Sunday"),
    (18, "Tuesday"),
    (7, "Wednesday"),
    (14, "Saturday"),
    (23, "Friday"),
    (8, "Thursday"),
]

def classify_access_period(hour: int, day: str) -> str:

    '''
    

'''
    working_hours = WORKING_HOURS_START <= hour < WORKING_HOURS_END
    after_hours = not working_hours
    is_weekend = day.strip().lower() in WEEKEND_DAYS
    
    for hour, day in access_events:
        clean_day = day.strip().lower()
        if is_weekend:
            access_period = "weekend access"

        elif after_hours and not is_weekend:
            access_period = "weekday after-hours access"

        else:
            access_period = "normal working hours"

    return access_period

assert classify_access_period(7, "Monday") == "weekday after-hours access"
assert classify_access_period(8, "Monday") == "normal working hours"
assert classify_access_period(17, "Monday") == "normal working hours"
assert classify_access_period(18, "Saturday") == "weekend access"
```

```python
# Part 4: assign_risk_band(ratio: float) -> str
# Risk bands
# Turn a peer ratio into a risk band:
# Ratio to peer median	Band
# Below 1.5	normal
# 1.5 up to (not including) 3	elevated
# 3 up to (not including) 5	high
# 5 or more	critical
# Write it with if/elif/else, store the result in risk_band, and test with ratios of 0.8, 1.5, 4.59 and 7.2.
# Then add a comment explaining what would go wrong if you checked >= 1.5 before >= 5.



def assign_risk_band(ratio: float) -> str:
    '''
assign risk to boundaries based on ratio computed    

'''
    if ratio < 1.5:
        risk_band = "normal"

    elif ratio < 3:
        risk_band = "elevated"

    elif ratio < 5:
        risk_band = "high"

    else:
        risk_band = "critical"

    return risk_band

assert assign_risk_band(1.49) == "normal"
assert assign_risk_band(1.5) == "elevated"
assert assign_risk_band(2.99) == "elevated"
assert assign_risk_band(5.0) == "critical"
```

```python
# Part 5: evaluate_employee_week(...)

peer_median_by_role = {
    "branch_adviser": 32,
    "contact_centre_agent": 210,
    "fraud_investigator": 180,
}

HIGH_VOLUME_ROLES = {"fraud_investigator", "contact_centre_agent"}
EXCESSIVE_ACCESS_MULTIPLIER = 3

employee_weeks = [
    {"employee_id": "E00417", "role": "branch_adviser", "unique_customers": 147},
    {"employee_id": "E00588", "role": "branch_adviser", "unique_customers": 29},
    {"employee_id": "E00731", "role": "fraud_investigator", "unique_customers": 640},
    {"employee_id": "E00902", "role": "contact_centre_agent", "unique_customers": 250},
    {"employee_id": "E01044", "role": "Branch Adviser", "unique_customers": 110},
]

def evaluate_employee_week(employee_week: list) -> tuple:
    '''
    Takes a list of employee record and return a tuple

    '''

    alerted_employee_ids = []
    for employee in employee_weeks:
        clean_role = "_".join(employee["role"].lower().split())
        peer_median = peer_median_by_role[clean_role]
        
        peer_median_ratio = employee["unique_customers"] / peer_median

        peer_median = peer_median_by_role[clean_role]
        ratio_to_peer = employee["unique_customers"] / peer_median

        # role aware rule
        is_excessive_access_alert = (
            peer_median_ratio > EXCESSIVE_ACCESS_MULTIPLIER 
            and clean_role not in HIGH_VOLUME_ROLES
        )

        if is_excessive_access_alert:
            alerted_employee_ids.append(employee["employee_id"])

        # Report and collect alerted id
        print(
            f"{employee['employee_id']} {clean_role:<22}"
            f"{peer_median_ratio:>5.2f}x alert: {is_excessive_access_alert}"
        )


    return f"Alerted: {alerted_employee_ids}"

alert = evaluate_employee_week(employee_weeks)
print(alert)
```
