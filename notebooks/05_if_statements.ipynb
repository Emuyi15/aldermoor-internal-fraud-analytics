```python
# Part 4: A median for any list
values = [28, 35, 31, 147, 29, 33, 40, 26, 32, 15]
values.sort()
print(values)

# median

if len(values) % 2 > 0:
    median = values[len(values)//2]

else:
    median = (values[(len(values)//2 - 1)] + values[len(values)//2]) / 2

print("_" * 50)
print("test 1")
print(f"median {median:.2f}")


values = [18, 22, 25, 30, 21, 27, 24, 29]
values.sort()
print(values)

# median

if len(values) % 2 > 0:
    median = values[len(values)//2]

else:
    median = (values[(len(values)//2 - 1)] + values[len(values)//2]) / 2

print("_" * 50)
print("test 2")
print(f"median {median:.2f}")
print(values)
```
